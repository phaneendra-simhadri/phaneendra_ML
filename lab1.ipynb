```python
def count(text):
  vowel = "aeiouAEIOU"
  vowels = 0
  consonants = 0

  for char in text:
    if char.isalpha():
      if char in vowel:
        vowels += 1
      else:
        consonants += 1
  return vowels,consonants

text = input("Enter the text: ")

vowels,consonants = count(text)
print("No.of vowels : ",vowels)
print("No.of consonants : ",consonants)
```

```python
def matrix_multiply(A, B):
    rows_A = len(A)
    cols_A = len(A[0])
    rows_B = len(B)
    cols_B = len(B[0])

    if cols_A != rows_B:
        return None

    result = []

    for i in range(rows_A):
        row = []
        for j in range(cols_B):
            total = 0
            for k in range(cols_A):
                total += A[i][k] * B[k][j]
            row.append(total)
        result.append(row)

    return result


# Main Program
rows_A = int(input("Enter the number of rows of Matrix A: "))
cols_A = int(input("Enter the number of columns of Matrix A: "))

A = []
print("Enter the elements of Matrix A:")
for i in range(rows_A):
    row = list(map(int, input().split()))
    A.append(row)

rows_B = int(input("Enter the number of rows of Matrix B: "))
cols_B = int(input("Enter the number of columns of Matrix B: "))

B = []
print("Enter the elements of Matrix B:")
for i in range(rows_B):
    row = list(map(int, input().split()))
    B.append(row)

product = matrix_multiply(A, B)

if product is None:
    print("Error: Matrix multiplication is not possible.")
else:
    print("Product of A and B:")
    for row in product:
        print(row)
```

```python
# need to check if there are duplicates repeating should cnt or not

def common(list1,list2):
  set1 = set(list1)
  count = 0

  for num in list2:
    if num in set1:
      count += 1
  return count

list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))

result = common(list1,list2)
print("No.of common elements : ",result)
```

```python
def transpose(A):
  rows = len(A)
  cols = len(A[0])

  result = [[0 for _ in range(rows)] for _ in range(cols)]

  for i in range(rows):
    for j in range(cols):
      result[j][i] = A[i][j]

  return result

rowsA = int(input("Enter the number of rows of Matrix A: "))

A = []

for i in range(rowsA):
  row = list(map(int, input().split()))
  A.append(row)

transA = transpose(A)
print("The transpose of A is :")
for row in transA:
  print(*row)
```
