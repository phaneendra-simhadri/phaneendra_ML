```python
#here its fact and non fact...so binary encoding is appropriate

import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from collections import Counter
from sklearn.model_selection import train_test_split


#load and apply binary encoding
def load_data(filepath):
  df = pd.read_csv(filepath)
  X_text = df['Doc'].values
  y = df['Label'].map({'Fact' : 0, 'Non-Fact': 1}).values
  return X_text, y


#using ts-idf vectors for text
def vectorize_text(X_text, max_features=3000):
    vectorizer = TfidfVectorizer(max_features=max_features, stop_words='english')
    X = vectorizer.fit_transform(X_text).toarray()
    return X, vectorizer


def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

def manhattan_distance(a, b):
    return np.sum(np.abs(a - b))

# cosine similarity ...dist = 1 - simi
def cosine_distance(a, b):
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    if norm_a == 0 or norm_b == 0:
        return 1.0
    return 1 - np.dot(a, b) / (norm_a * norm_b)


#sorting algorithms
def bubble_sort(arr):
  n = len(arr)
  for i in range(n):
    for j in range(0, n-i-1):
      if arr[j][0] > arr[j+1][0]:
        arr[j], arr[j+1] = arr[j+1], arr[j]
  return arr

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j][0] < arr[min_idx][0]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i-1
        while j >=0 and key[0] < arr[j][0]:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr


SORT_ALGOS = {
    'bubble': bubble_sort,
    'selection': selection_sort,
    'insertion': insertion_sort
}


def compute_distances(X_train, test_point, metric='euclidean'):
    if metric == 'euclidean':
        dist_func = euclidean_distance
    elif metric == 'manhattan':
        dist_func = manhattan_distance
    elif metric == 'cosine':
        dist_func = cosine_distance
    else:
        raise ValueError("Unsupported metric")

    distances = []
    for i, train_point in enumerate(X_train):
        d = dist_func(test_point, train_point)
        distances.append((d, i))
    return distances


def identify_neighbors(distances, k, sort_algo='insertion'):
    sorted_dist = SORT_ALGOS[sort_algo](distances.copy())
    neighbors = sorted_dist[:k]
    if len(sorted_dist) > k and sorted_dist[k-1][0] == sorted_dist[k][0]:
        while k < len(sorted_dist) and sorted_dist[k-1][0] == sorted_dist[k][0]:
            neighbors.append(sorted_dist[k])
            k += 1
    return neighbors

def majority_vote(neighbor_indices, y_train):
    labels = [y_train[idx] for _, idx in neighbor_indices]
    counter = Counter(labels)
    max_count = max(counter.values())
    candidates = [cls for cls, cnt in counter.items() if cnt == max_count]
    return min(candidates)


 X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)





if __name__ == "__main__":
  #load
  filepath = "/content/Fact-NonFact_dataset1300 (version 1).xlsb.csv"
  X_text, y = load_data(filepath)


  # Vectorize text
  X, vectorizer = vectorize_text(X_text, max_features=3000)
  print(f"Feature dimensions: {X.shape[1]}")



```
